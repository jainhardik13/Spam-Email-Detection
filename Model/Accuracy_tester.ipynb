```python
# 1️⃣ Load model and vectorizer
model = pickle.load(open("spam_model1.pkl", "rb"))
vectorizer = pickle.load(open("vectorizer1.pkl", "rb"))
```

```python
# 2️⃣ Load NEW unseen test dataset
test_df = pd.read_excel("sms_dataset_500_labeled_full.xlsx")
```

```python
# Separate columns
X_test = test_df["text"]
y_test = test_df["label"]
```

```python
# Transform
X_test_vectorized = vectorizer.transform(X_test)
```

```python
# Predict
y_pred = model.predict(X_test_vectorized)
```

```python
# Metrics
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
